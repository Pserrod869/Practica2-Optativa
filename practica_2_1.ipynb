```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
texto = "me gustan los videojuegos"
resultado = clasificador(texto)
print(resultado)
```

Usando este modelo vamos hacer una mini aplicacion que vaya en colab que disponga de una lista negra de palabras que siempre vamos a excluir y vamos a excluir los menasejs que se consideren como negativos ( frase que contenga una palabra mala lafrase es eliminada) y tiene que poner porque no es aceptada la frase o palabra (Hay que hacer una lista 10 frases y 10 palabras)

```python
import re

lista_negra = [
    "odio", "basura", "tonto", "estúpido", "asco",
    "maldito", "peor", "horrible", "pesadilla", "feo"
]

frases_prueba = [
    "Me encantan los videojuegos y pasar tiempo con mis amigos.",
    "Este producto es una completa basura, no lo compren.",
    "Hoy es un día increíble para salir a pasear.",
    "Qué servicio tan horrible, me atendieron súper mal.",
    "Creo que el clima de hoy está bastante templado y normal.",
    "No me gusta para nada cómo me trataste ayer.",
    "Es un tonto si piensa que va a ganar el partido.",
    "¡Qué asco de comida, la verdad no pude terminarla!",
    "Muchas gracias por toda la ayuda que me brindaste hoy.",
    "Ese ruido de la calle es una pesadilla constante."
]

def procesar_frases(frases, palabras_prohibidas, clasificador_sentimiento):
    print("=== INICIANDO ANÁLISIS DE FRASES ===\n")

    for i, frase in enumerate(frases, 1):
        print(f"Frase {i}: \"{frase}\"")

        frase_normalizada = re.sub(r'[^\w\s]', '', frase.lower())
        palabras_frase = frase_normalizada.split()

        palabras_detectadas = [palabra for palabra in palabras_prohibidas if palabra in palabras_frase]

        if palabras_detectadas:
            print(f"❌ RECHAZADA: Contiene palabra(s) prohibida(s) de la lista negra: {palabras_detectadas}\n")
            continue

        prediccion = clasificador_sentimiento(frase)[0]
        sentimiento = prediccion['label']
        score = prediccion['score']

        if sentimiento == 'NEG':
            print(f"❌ RECHAZADA: El modelo detectó un sentimiento NEGATIVO (Confianza: {score:.2%})\n")
        else:
            print(f"✅ ACEPTADA: Sentimiento clasificado como {sentimiento} (Confianza: {score:.2%})\n")

procesar_frases(frases_prueba, lista_negra, clasificador)
```
